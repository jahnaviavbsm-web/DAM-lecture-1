```python
#Byjus Case
import pandas as pd
import numpy as np

df = pd.read_excel("/content/Lecture2_data.xlsx")
```

```python
#Data Description
print(df.head)
print(df.shape)
```

```python
df.info()
```

```python
#Top 10 profit/loss product
df["Total_Cost_Lakh"] = (
    df["Marketing_Spend_Lakh"]
    + df["Content_Cost_Lakh"]
    + df["Salary_Cost_Lakh"]
    + df["Other_Cost_Lakh"]
    )

df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total_Cost_Lakh"]

print(df[["Product", "Revenue_Lakh", "Total_Cost_Lakh", "Profit_Lakh"]].head(10))
```

```python
#Missing Value
print(df.isna().sum())
```

```python
# Decision making structures
#If Statements
rating = df["Customer_Rating"].mean()

if rating >= 4:
    print("Customer satisfaction is good")
```

```python
# Decision making structures
#If Statements
revenue = df["Revenue_Lakh"].mean()

if revenue >= 200:
    print("Company is profitable")
```

```python
# Decision making structures
#If else
revenue = df["Revenue_Lakh"].mean()

if revenue >= 300:
    print("Average revenue is high")
else:
    print("Average revenue is low")
```

```python
#Desicion making structure
#Nested IF Statement in Python
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
    print("Revenue is High")

    if rating >= 4:
        print("Customer satisfaction is also good")
    else:
        print("Customer satisfaction needs improvement")
else:
    print("Revenue is low")
```

```python
# Find the average customer rating
rating = df["Customer_Rating"].mean()

#Make a decision based ont he rating
if rating >= 4.5:
  print("Excellent")

elif rating >= 4.0:
  print("Good")

elif rating >=3.0:
  print("Average")

else:
  print("Needs improvement")
```

```python
# For loop
for rating in df["Customer_Rating"]:

    if rating >= 4:
       print("Good rating")

    else:
      print("Needs Improvement")
```

```python
# for loop with range() iloc= integer locaation

for i in range(10):
  print(df['Product'].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#WHILE Loop in Python
#While loop and i,j is variable to track
i = 0

while i < 5:
  revenue = df["Revenue_Lakh"].iloc[i]

  if revenue >= 300:
    print("High Revenue", revenue)
  else:
    print("Low Revenue", revenue)

  i = i + 1
```
